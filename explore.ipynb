```python
adv = df1[["PLAYER_ID","W_PCT",'TS_PCT','USG_PCT','PIE','FG_PCT','NET_RATING','AST_PCT','REB_PCT']]
```

```python
merged = df.merge(adv, on='PLAYER_ID',how='left')
```

```python
merged.head()
```

```python
(merged['MIN'] > 1000).head()
```

```python
merged.filter(merged['MIN']>1000)
```

```python
merged = merged[merged['MIN'] > 1000].copy()
```

```python
merged.shape
```

```python
merged['FG3A_RATE'] = merged['FG3A'] / merged['FGA']
merged[['PLAYER_NAME', 'FG3A_RATE']].head(10)
```

```python
merged['FG3A_RATE'].describe()
```

```python
merged['PTS_PER_SHOT']=merged['PTS']/merged['FGA']
```

```python
merged[['PLAYER_NAME', 'PTS_PER_SHOT']].head(10)
```

```python
merged.columns.tolist()
```

```python
merged['STL_per36']  = merged['STL']  / merged['MIN'] * 36
merged['BLK_per36']  = merged['BLK']  / merged['MIN'] * 36
merged['OREB_per36'] = merged['OREB'] / merged['MIN'] * 36
merged['DREB_per36'] = merged['DREB'] / merged['MIN'] * 36
merged['PFD_per36']  = merged['PFD']  / merged['MIN'] * 36
merged['TOV_per36']  = merged['TOV']  / merged['MIN'] * 36

style_features = [
    'USG_PCT', 'AST_PCT', 'TS_PCT', 'FG3A_RATE',
    'STL_per36', 'BLK_per36', 'OREB_per36', 'DREB_per36', 'PFD_per36', 'TOV_per36'
]
```

```python
merged[style_features].describe()

merged[style_features].info()
```

```python
merged.to_csv('nba_clean.csv', index=False)
```

```python
print(merged.shape)
print(merged[style_features].describe())
print(merged[style_features].isna().sum()) 
```

```python
merged[style_features].corr()
```

```python
from sklearn.preprocessing import StandardScaler
X = merged[style_features]
X_scaled = StandardScaler().fit_transform(X)
```

```python
pd.DataFrame(X_scaled, columns=style_features).describe()
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

```python
v = []
inertias = []
for k in range(1,11):
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    inertia = km.fit(X_scaled).inertia_
    v.append(k)
    inertias.append(inertia)
    

plt.plot(v,inertias)
    
```

```python
from sklearn.metrics import silhouette_score
for k in range(2, 11):
    labels = KMeans(n_clusters=k, random_state=42, n_init=10).fit_predict(X_scaled)
    print(f"k={k}: silhouette={silhouette_score(X_scaled, labels):.3f}") 
```

```python
km = KMeans(n_clusters=3, random_state=42, n_init=10)
merged['cluster'] = km.fit_predict(X_scaled)
```

```python
merged['cluster'].value_counts()
```

```python
merged.groupby('cluster')[style_features].mean()
```

```python
for c in [0, 1, 2]:
    print(f"\n--- Cluster {c} ---")
    print(merged[merged['cluster'] == c]['PLAYER_NAME'].sample(8).tolist())
```

```python
archetype_names = {
    0: 'Primary Creator',
    1: '3-and-D Wing',
    2: 'Interior Big'
}
merged['archetype'] = merged['cluster'].map(archetype_names)
merged['archetype'].value_counts()
```
